```python
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun

search = DuckDuckGoSearchRun()

@tool
def web_search(query: str) -> str:
    """
    Search the web for up-to-date information.

    Use this tool when:
    - current events are needed
    - recent facts are needed
    - latest company hiring/news is needed

    If no results are found, return NO_RESULTS_FOUND.
    """

    try:
        result = search.run(query)

        if not result or len(result.strip()) == 0:
            return "NO_RESULTS_FOUND"

        return result[:1000]

    except Exception as e:
        return f"SEARCH_ERROR: {str(e)}"
```

```python
print(web_search.invoke({
    "query": "TCS hiring 2026"
}))
```

```python
from langgraph.prebuilt import create_react_agent
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

agent = create_react_agent(
    llm,
    tools=[web_search]
)

print("Agent created successfully.")
```

```python
result = agent.invoke(
    {
        "messages": [
            ("user", "What is TCS's 2026 hiring quota?")
        ]
    },
    config={
        "recursion_limit": 10
    }
)

for i, m in enumerate(result["messages"]):

    print(f"\n[{i}] {type(m).__name__}")

    if hasattr(m, "content"):
        print("Content:")
        print(str(m.content)[:500])

    if hasattr(m, "tool_calls") and m.tool_calls:
        print("Tool Calls:")
        print(m.tool_calls)
```

```python
import requests
from bs4 import BeautifulSoup
from langchain_core.tools import tool

@tool
def jd_fetcher(url: str) -> str:
    """
    Fetch a job description from a URL and return clean plain text.

    Use ONLY when the user provides a job posting URL.

    Returns:
    - cleaned text from webpage
    - OR ERROR string if fetch fails
    """

    try:
        r = requests.get(
            url,
            headers={"User-Agent": "Mozilla/5.0"},
            timeout=10
        )

        r.raise_for_status()

        soup = BeautifulSoup(r.text, "html.parser")

        for tag in soup(["script", "style"]):
            tag.decompose()

        text = soup.get_text(
            separator="\n",
            strip=True
        )

        return text[:4000]

    except Exception as e:
        return f"ERROR: failed to fetch URL — {str(e)}"
```

```python
print(jd_fetcher.invoke({
    "url": "https://careers.tcs.com"
})[:500])
```

```python
@tool
def skills_gap(
    student_skills: str,
    must_have_skills: str
) -> str:
    """
    Compare student skills to job required skills.

    Returns missing skills.
    """

    student = set(
        s.strip().lower()
        for s in student_skills.split(",")
        if s.strip()
    )

    required = set(
        s.strip().lower()
        for s in must_have_skills.split(",")
        if s.strip()
    )

    missing = sorted(required - student)

    return ", ".join(missing) if missing else "none"
```

```python
print(skills_gap.invoke({
    "student_skills": "Python, Java, SQL",
    "must_have_skills": "Python, Java, SQL, AWS, Spring Boot"
}))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)

@tool
def answer_scorer(
    question: str,
    answer: str
) -> str:
    """
    Score interview answer from 1-10 with rationale.
    """

    prompt = f"""
    Score this interview answer from 1-10.

    Question:
    {question}

    Answer:
    {answer}

    Return format:
    Score: X/10
    Rationale: <short reason>
    """

    try:
        return llm.invoke(prompt).content

    except Exception as e:
        return f"ERROR: {str(e)}"
```

```python
print(answer_scorer.invoke({
    "question": "Why TCS Digital?",
    "answer": "Because TCS is big and they pay well."
}))
```

```python
from langgraph.prebuilt import create_react_agent

tools = [
    jd_fetcher,
    skills_gap,
    answer_scorer
]

agent = create_react_agent(
    llm,
    tools=tools
)

print(f"Agent created with {len(tools)} tools.")
```

```python
profiles = [
    {
        "name": "Ravi Kumar",
        "branch": "CSE",
        "cgpa": 8.2,
        "skills": ["Python", "Java", "SQL"],
        "target_company": "TCS"
    },
    {
        "name": "Sneha Reddy",
        "branch": "ECE",
        "cgpa": 8.8,
        "skills": ["C", "Python", "DBMS"],
        "target_company": "Cognizant"
    },
    {
        "name": "Arun Pillai",
        "branch": "IT",
        "cgpa": 9.1,
        "skills": ["Python", "AWS", "DSA"],
        "target_company": "Amazon"
    }
]
```

```python
for i, p in enumerate(profiles):

    print("\n" + "="*70)

    print(
        f"Student {i+1}: "
        f"{p['name']} → {p['target_company']}"
    )

    print("="*70)

    msg = (
        f"I am {p['name']}, "
        f"B.Tech {p['branch']} "
        f"CGPA {p['cgpa']}, "
        f"skills: {', '.join(p['skills'])}. "
        f"Target company: {p['target_company']}. "
        f"Give me mock interview questions, "
        f"score one answer, "
        f"and tell skill gaps."
    )

    result = agent.invoke(
        {
            "messages": [
                ("user", msg)
            ]
        },
        config={
            "recursion_limit": 10
        }
    )

    for j, m in enumerate(result["messages"]):

        print(f"\n[{j}] {type(m).__name__}")

        if hasattr(m, "content"):
            print(str(m.content)[:500])

        if hasattr(m, "tool_calls") and m.tool_calls:
            print(m.tool_calls)
```
